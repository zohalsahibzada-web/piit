```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/selva86/datasets/master/mtcars.csv"
mtcars = pd.read_csv(url)

mtcars.head()
```

```python
mtcars["vs"] = mtcars["vs"].replace({0: "v-shaped", 1: "straight"})
mtcars["am"] = mtcars["am"].replace({0: "automatic", 1: "manual"})
```
